# Notebook 11 — Task C: GenAI Explanation Layer

**Two components:**

1. **Executive Briefing Generator** — Automated narrative per brand: volume outlook, market share direction, attribution drivers, territory spotlight, strategic signal.

2. **Ask-Your-Forecast Copilot** — Q&A interface over the forecast outputs. Works offline using structured knowledge base. Automatically upgrades to Claude when `ANTHROPIC_API_KEY` is set.

**Why this matters:** The model produces 3,840 rows of numbers. GenAI turns those numbers into decisions — surfacing the *why* behind every share movement and flagging where action is needed.

**Portability:** Runs on any system with no API key required. Claude enhancement is optional and additive.

## Step 6 — Ask-Your-Forecast Copilot

The copilot answers natural-language questions using the forecast knowledge base. No API key needed for these answers — all grounded in actual model outputs.

```python
from run_genai_narrative import ask_copilot, build_knowledge_base, BRAND_META
import pandas as pd
import numpy as np

# Reload all data (same as script)
ROOT = Path('..').resolve()
import sys
sys.path.insert(0, str(ROOT / '03_scripts'))

# Load the pre-built knowledge base
kb = json.loads((OUT / 'copilot_knowledge_base.json').read_text())
print('Copilot knowledge base loaded.')
print(f'Brands: {kb["portfolio"]["brands"]}')
```

### Interactive Q&A Demo

```python
questions = [
    'Which brands are gaining market share?',
    'Why is Ocretiva losing share?',
    'What is Phesgrox forecast for H1 2025?',
    'What is Hemvia market share forecast?',
    'How accurate is our model vs TM1?',
    'What is the biggest risk in the portfolio?',
    'Why is Vabyseal gaining share?',
    'What is Perjenta losing share?',
]

print('=' * 70)
print('ASK-YOUR-FORECAST COPILOT')
print('=' * 70)
for q in questions:
    answer = ask_copilot(q, kb)
    print(f'\nQ: {q}')
    print(f'A: {answer}')
    print('-' * 50)
```
